```python
# ============================================================
# MASTER NOTEBOOK: RUN ALL OTHER NOTEBOOKS AUTOMATICALLY
# ============================================================
# pip install papermill
import papermill as pm
import os

# List your notebooks in the order you want them run
NOTEBOOKS_TO_RUN = [
    "baseline-mobileV3-classifierOnly.ipynb",
    "baseline-mobileV3-Partial.ipynb",
    "baseline-mobileV3-Full.ipynb",
    "baseline-shufflenetV2-Full.ipynb",
    "baseline-shufflenetV2-Partial.ipynb",
    "baseline-shufflenetV2-classifierOnly.ipynb",
    "baseline-efficientnet_b0-Full.ipynb",
    "baseline-efficientnet_b0-Partial.ipynb",
    "baseline-efficientnet_b0-classifierOnly.ipynb",
    "LightcustomCNN.ipynb",
    
    
    
    # "comparison.ipynb",
]

OUTPUT_DIR = "executed_notebooks"
os.makedirs(OUTPUT_DIR, exist_ok=True)

for nb in NOTEBOOKS_TO_RUN:
    print(f"\n{'='*60}\nRunning: {nb}\n{'='*60}")
    output_path = os.path.join(OUTPUT_DIR, f"executed_{nb}")
    try:
        pm.execute_notebook(
            input_path=nb,
            output_path=output_path,
            kernel_name="cvpr",   # match your cvpr conda kernel name
            log_output=True,
        )
        print(f"Completed: {nb} -> saved to {output_path}")
    except Exception as e:
        print(f"FAILED: {nb} — {e}")
        # continue to next notebook instead of stopping everything
        continue

print("\n\n=== ALL NOTEBOOKS PROCESSED ===")
```
